```python
import chess
import chess.engine

# Khởi động engine
engine = chess.engine.SimpleEngine.popen_uci("/usr/games/stockfish")
board = chess.Board()

# Hỏi người chơi có muốn đi trước không
first = input("Bạn có muốn đi trước không? (1 = Có, 0 = Không): ")
while first not in ['0', '1']:
    first = input("Vui lòng nhập 1 hoặc 0: ")
human_is_white = first == '1'

# Nếu máy đi trước
if not human_is_white:
    print("Stockfish đi trước...")
    result = engine.play(board, chess.engine.Limit(time=1.0))
    print("Stockfish đánh:", result.move)
    board.push(result.move)

# Vòng chơi
while not board.is_game_over():
    print("\nBàn cờ hiện tại:")
    print(board)

    if board.turn == human_is_white:
        move_input = input("Bạn đi (ví dụ e2e4, hoặc 'undo' để đi lại): ")
        if move_input.lower() == "undo":
            if len(board.move_stack) >= 2:
                board.pop()  # undo nước máy
                board.pop()  # undo nước người
                print("⏪ Đã quay lại 1 lượt.")
            elif len(board.move_stack) == 1:
                board.pop()
                print("⏪ Đã quay lại nước đầu.")
            else:
                print("⚠️ Không có nước nào để đi lại.")
            continue
        try:
            move = chess.Move.from_uci(move_input)
            if move in board.legal_moves:
                board.push(move)
            else:
                print("❌ Nước đi không hợp lệ.")
        except:
            print("❌ Sai định dạng. Hãy nhập theo UCI (ví dụ: e2e4)")
    else:
        print("Stockfish đang suy nghĩ...")
        result = engine.play(board, chess.engine.Limit(time=1.0))
        print("Stockfish đánh:", result.move)
        board.push(result.move)

# Kết thúc ván
print("\n🏁 Ván cờ kết thúc! Kết quả:", board.result())
engine.quit()
```
